# 04 — What "MVP" means for this training stage

The stage-4 training notebooks are deliberately minimal. This notebook is not part of the pipeline — it has no inputs or outputs of its own — it exists purely to write down, in one place, what a stage-4 training notebook is supposed to prove and what it intentionally leaves out, so the scope stays explicit instead of implied.

## What the MVP is supposed to do

The goal of stage 4 is to answer one question as cheaply as possible: **given the feature artifacts produced by stages 1–3, is there a learnable signal at all, and how much better than a naive guess can a simple model get?**

Concretely, a stage-4 notebook:

- loads the pre-built train/test feature artifacts for one configured target,
- restricts both splits to the same eligibility cohort (complete targets, complete predictors),
- fits exactly one candidate model with a hand-picked, fixed configuration,
- scores that one model once on the held-out test cohort, and
- displays aggregate and per-horizon error metrics plus a small prediction preview, inline, in the notebook.

That's it. Each notebook is one reference point. Everything downstream — is this candidate worth pursuing, is the pipeline plumbing sound end-to-end — can be answered by reading a handful of such notebooks' output side by side.

## What it deliberately does not do

- **No hyperparameter tuning or cross-validation.** Any hyperparameters are hardcoded guesses, not values chosen by search. There is no CV loop and no validation split distinct from the test split.
- **No experiment tracking.** Metrics are printed and displayed, not logged anywhere (no run history, no comparison store). Re-running the notebook overwrites the only copy of the result — the previous run's numbers exist nowhere but a person's memory or a git-tracked notebook output.
- **No model artifacts.** The fitted model lives only in the notebook's kernel memory for the duration of the run. Nothing is saved, versioned, or made loadable by another process.
- **No model selection.** Candidates are not compared programmatically; a human reads each notebook's printed metrics and decides which (if any) looks promising.
- **Single target only.** One configured target (e.g. one station) is hardcoded. Nothing here says whether the same approach, or the same configuration, works anywhere else.
- **Single train/test split.** One chronological cut, one evaluation pass. The reported error is one sample, not a distribution — there is no rolling-origin backtest to show how performance varies across time windows.
- **Point forecasts only.** No prediction intervals or uncertainty estimates, just a single predicted value per horizon.
- **No production concerns.** No serving path, no scheduled retraining, no drift monitoring, no data-quality alerting. Running the notebook by hand *is* the whole operational story.

None of this is accidental — it's the cost of getting a first, trustworthy answer quickly. Each bullet above is a candidate for later work, not a defect to silently fix.
